```python
# ─────────────────────────────────────────
# DBSCAN — Density-Based Spatial Clustering of Applications with Noise
# ─────────────────────────────────────────
import math

# Points 2D à regrouper
points = [
    (1,1),(1,2),(2,1),(2,2),       # Cluster 1
    (8,8),(8,9),(9,8),(9,9),       # Cluster 2
    (5,5)                          # Bruit (point isolé)
]

# Hyper-paramètres
eps     = 2.0   # Rayon de voisinage
min_pts = 2     # Nombre minimum de voisins pour être un "core point"

# Distance euclidienne entre deux points
def distance(a, b):
    return math.sqrt((a[0]-b[0])**2 + (a[1]-b[1])**2)

# Trouver tous les voisins dans le rayon eps
def voisins(p, pts):
    return [i for i, q in enumerate(pts) if distance(p, q) <= eps]

# Algorithme DBSCAN
labels = [-1] * len(points)   # -1 = non visité / bruit
cluster_id = 0

for i, p in enumerate(points):
    if labels[i] != -1: continue   # Déjà traité

    vois = voisins(p, points)

    if len(vois) < min_pts:        # Pas assez de voisins → bruit
        labels[i] = 0
        continue

    # Nouveau cluster : étendre depuis ce core point
    cluster_id += 1
    labels[i] = cluster_id
    file = list(vois)              # File d'expansion

    while file:
        j = file.pop()
        if labels[j] == 0: labels[j] = cluster_id   # Bruit → frontière
        if labels[j] != -1: continue                 # Déjà assigné
        labels[j] = cluster_id
        vois_j = voisins(points[j], points)
        if len(vois_j) >= min_pts:
            file.extend(vois_j)   # Étendre si core point

# Affichage
print("Point        → Cluster")
for i, p in enumerate(points):
    etiquette = f"Cluster {labels[i]}" if labels[i] > 0 else "Bruit"
    print(f"  {str(p):<12} → {etiquette}")
```
