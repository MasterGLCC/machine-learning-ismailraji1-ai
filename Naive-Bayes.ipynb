```python
# ─────────────────────────────────────────
# NAIVE BAYES — Classifieur probabiliste de Bayes
# ─────────────────────────────────────────
from collections import defaultdict, Counter

# Données d'entraînement : [Météo, Vent] → Classe
train = [
    (["Soleil","Faible"], "Oui"), (["Soleil","Fort"], "Non"),
    (["Nuage","Faible"],  "Oui"), (["Pluie","Faible"],  "Oui"),
    (["Pluie","Fort"],    "Non"), (["Nuage","Fort"],    "Oui"),
    (["Soleil","Fort"],   "Non"), (["Pluie","Faible"],  "Oui"),
]

classes = ["Oui", "Non"]

# Compter les occurrences par classe et par (feature, valeur)
n_total = len(train)
count_classe = Counter(c for _, c in train)       # Nb d'exemples par classe

# count_feat[classe][col][valeur] = nb d'occurrences
count_feat = defaultdict(lambda: defaultdict(Counter))
for feats, cls in train:
    for col, val in enumerate(feats):
        count_feat[cls][col][val] += 1

# Probabilité P(classe)
def p_classe(cls):
    return count_classe[cls] / n_total

# Probabilité P(valeur | classe) avec lissage de Laplace (évite les 0)
def p_feature_given_classe(val, col, cls):
    vocab = set(r[col] for r, _ in train)       # Valeurs uniques de cette colonne
    num   = count_feat[cls][col][val] + 1       # +1 = lissage Laplace
    den   = count_classe[cls] + len(vocab)
    return num / den

# Prédiction : choisir la classe avec la plus haute probabilité
def predict(x):
    scores = {}
    for cls in classes:
        score = p_classe(cls)                  # Prior P(C)
        for col, val in enumerate(x):
            score *= p_feature_given_classe(val, col, cls)  # × P(xi|C)
        scores[cls] = score
    return max(scores, key=scores.get)         # argmax

# Test
tests = [["Soleil","Fort"], ["Nuage","Faible"], ["Pluie","Fort"]]
print("Naive Bayes — Prédictions\n")
for t in tests:
    print(f"  {t} → Jouer ? {predict(t)}")
```
