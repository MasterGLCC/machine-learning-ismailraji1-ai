```python
# ─────────────────────────────────────────
# Q-LEARNING — Apprentissage par renforcement
# ─────────────────────────────────────────
import random

# États possibles (salles d'un labyrinthe simple)
states  = ["A", "B", "C", "D"]

# Actions possibles depuis chaque état (transitions)
actions = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["D"],
    "D": []           # État terminal (but atteint)
}

# Récompenses : atteindre D = +10, sinon -1
rewards = {("A","B"): -1, ("A","C"): -1,
           ("B","A"): -1, ("B","D"): 10,
           ("C","D"): 10}

# Table Q initialisée à 0 — Q[état][action] = valeur apprise
Q = {s: {a: 0 for a in actions[s]} for s in states if actions[s]}

# Hyper-paramètres
alpha   = 0.8   # Taux d'apprentissage
gamma   = 0.9   # Facteur d'actualisation (importance du futur)
epsilon = 0.3   # Probabilité d'exploration aléatoire

# Un épisode d'apprentissage
def episode():
    state = random.choice(["A","B","C"])   # Départ aléatoire (pas D)
    for _ in range(20):
        if state == "D" or not actions[state]: break
        # Politique ε-greedy : explorer ou exploiter
        if random.random() < epsilon:
            action = random.choice(actions[state])   # Explorer
        else:
            action = max(Q[state], key=Q[state].get)  # Exploiter

        reward  = rewards.get((state, action), 0)
        suivant = action                             # L'action = prochain état

        # Valeur future maximale (Bellman)
        q_futur = max(Q[suivant].values()) if suivant in Q and Q[suivant] else 0

        # Mise à jour Q : Q(s,a) ← Q(s,a) + α[r + γ·max Q(s',a') - Q(s,a)]
        Q[state][action] += alpha * (reward + gamma * q_futur - Q[state][action])
        state = suivant

# Lancer 100 épisodes d'apprentissage
for _ in range(100):
    episode()

print("Table Q apprise :")
for s, acs in Q.items():
    print(f"  État {s} : {acs}")
# Le meilleur chemin : A→B→D ou A→C→D (Q[B][D] et Q[C][D] doivent être les plus élevés)
```
