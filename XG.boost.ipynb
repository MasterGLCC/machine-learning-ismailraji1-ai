```python
# ─────────────────────────────────────────
# XGBoost — Gradient Boosting simplifié from scratch
# ─────────────────────────────────────────

# Données : taille (cm) → poids (kg)
X = [150, 160, 170, 180, 190]
y = [50,  60,  70,  80,  90]   # Vraies valeurs

# Hyper-paramètres
eta        = 0.3   # Taux d'apprentissage (learning rate)
n_arbres   = 5     # Nombre de boosting rounds
prediction = [60.0] * len(X)  # Prédiction initiale = moyenne

# Fonction de perte : Mean Squared Error
def mse(y_true, y_pred):
    return sum((a-b)**2 for a,b in zip(y_true,y_pred)) / len(y_true)

# Un "arbre" simplifié : prédit le résidu moyen (stumps)
def arbre_residus(residus):
    return sum(residus) / len(residus)  # Feuille unique = moyenne des résidus

# Boucle de Boosting
# Chaque itération corrige les erreurs des précédentes
print("Boosting XGBoost simplifié\n")
for t in range(n_arbres):
    # 1. Calculer les résidus = erreur du modèle actuel
    residus = [y[i] - prediction[i] for i in range(len(X))]

    # 2. Ajuster un arbre sur les résidus
    correction = arbre_residus(residus)

    # 3. Mettre à jour les prédictions avec la correction pondérée
    prediction = [prediction[i] + eta * correction for i in range(len(X))]

    print(f"  Round {t+1} | MSE = {mse(y, prediction):.2f} | Correction = {correction:.2f}")

print("\nPrédictions finales :", [round(p,1) for p in prediction])
print("Vraies valeurs      :", y)
```
