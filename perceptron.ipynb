```python
import math  # pour utiliser l'exponentielle 
 
# 1- Entrées 
x1 = 3 
x2 = 7 
 
# 2- Poids 
# 2 poids du neurone (un pour chaque entrée)  
w1 = 0.4     
w2 = 0.6 
 
# 3- Biais 
# valeur ajoutée à la somme pondérée pour ajuster la sortie du neurone 
b = -2     
 
# 4- Somme pondérée 
z = (x1 * w1) + (x2 * w2) + b    # L'équation d'un perceptron 
 
# 5- Fonction d'activation Sigmoïde  
# transforme une valeur en probabilité entre 0 et 1 
def sigmoid(x): 
    return 1 / (1 + math.exp(-x)) 
 
# 6- Sortie du neurone 
# Application de la fonction Sigmoid à la somme pondérée 
y = sigmoid(z)       
print("somme pondérée  =", z) 
print("σ(z) =", y) 
 
# 7- Classification 
print("pour point (3,7):") 
classe = 1 if y >= 0.5 else 0 
print("Classe =", classe)
```
