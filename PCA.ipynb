```python
# ─────────────────────────────────────────
# PCA — Analyse en Composantes Principales (from scratch)
# ─────────────────────────────────────────

# Données 3D : on veut les réduire en 2D
X = [
    [2.5, 2.4, 0.5],
    [0.5, 0.7, 1.0],
    [2.2, 2.9, 0.8],
    [1.9, 2.2, 0.3],
    [3.1, 3.0, 1.2],
    [2.3, 2.7, 0.6],
]
n, d = len(X), len(X[0])

# 1. Centrer les données : soustraire la moyenne de chaque colonne
moyennes = [sum(X[i][j] for i in range(n)) / n for j in range(d)]
Xc = [[X[i][j] - moyennes[j] for j in range(d)] for i in range(n)]

# 2. Calculer la matrice de covariance (d×d)
def cov_matrix(data):
    n, d = len(data), len(data[0])
    C = [[0]*d for _ in range(d)]
    for j in range(d):
        for k in range(d):
            C[j][k] = sum(data[i][j]*data[i][k] for i in range(n)) / (n-1)
    return C

C = cov_matrix(Xc)
print("Matrice de covariance :")
for row in C: print("  ", [round(v,3) for v in row])

# 3. Puissance itérée pour trouver la 1ère composante principale (vecteur propre dominant)
import math
def normalise(v): n=math.sqrt(sum(x**2 for x in v)); return [x/n for x in v]
def produit_mat_vec(M, v): return [sum(M[i][j]*v[j] for j in range(len(v))) for i in range(len(M))]

vec = normalise([1, 1, 1])        # Vecteur initial aléatoire
for _ in range(200):              # Itérations de la méthode de la puissance
    vec = normalise(produit_mat_vec(C, vec))

# Valeur propre associée (λ = v^T · C · v)
Cv = produit_mat_vec(C, vec)
valeur_propre = sum(vec[j]*Cv[j] for j in range(d))

# 4. Projeter chaque point sur la 1ère composante principale
projections = [sum(Xc[i][j]*vec[j] for j in range(d)) for i in range(n)]

print(f"\n1ère composante principale : {[round(v,3) for v in vec]}")
print(f"Variance expliquée (λ)     : {round(valeur_propre,4)}")
print(f"Projections (données 1D)   : {[round(p,3) for p in projections]}")
```
